```python
def temp_converter(temp,unit):
    if unit=="C":
        temprature=temp*9/5+32
        print(temprature)
    elif unit=="F":
        temprature=(temp-32)*9/5
        print(temprature)
```

```python

temp_converter(55,"F")
```

```python
def check_password(password):
    special_char="!@#$%^&*"
    if len(password)<8:
        print("entered password is weak")
    elif not any(char.lower() for char in password):
        print("entered password is weak")
    elif not any(char.upper() for char in password):
        print("entered password is weak")
    elif not any(char.isdigit() for char in password):
        print("entered password is weak")
    elif not any(char in special_char for char in password):
        print("entered password is weak")
    else:
        print("entered password is strong")
```

```python
check_password("mbykhaskvkkk")
```

```python
def palindrome(string):
    string=string.lower().replace(" ","")
    if string == string[::-1]:
        print("Its palindrome")
    else:
        print("Not a palindrome")
```

```python
palindrome("aba")
```
