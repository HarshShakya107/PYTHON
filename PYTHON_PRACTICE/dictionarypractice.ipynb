names = ["A", "B", "C"]
scores = [80, 90, 75]

create:

{"A": 80, "B": 90, "C": 75}

```python
names = ["A", "B", "C"]
scores = [80, 90, 75]
ans={name:scores for name,scores in zip(names,scores)}
print(ans)
```

Count the frequency of every element in:
["apple", "banana", "apple", "orange", "banana", "apple"]

```python
lists=["apple", "banana", "apple", "orange", "banana", "apple"]
ans={}
for x in lists:
    if x not in ans:
        ans.update({x:1})
    elif x in ans:
        value=ans[x]
        ans.update({x:value+1})
print(ans)
```

Given two dictionaries, merge them into a new dictionary. If the same key exists in both, add their values.

```python
a={"name":"Harsh","age":19}
b={"city":"Sehore","age":1}
c=a.copy()
for key,value in b.items():
    if key in c:
        c[key]+=value
    else:
        c[key]=value
print(c)

```

Group words by their first letter:
["apple", "ant", "banana", "ball", "cat"]

Expected structure:

{
    "a": ["apple", "ant"],
    "b": ["banana", "ball"],
    "c": ["cat"]
}

```python
lists=["apple", "ant", "banana", "ball", "cat"]
ans={}
for word in lists:
    letter=word[0]
    if letter in ans:
        ans[letter].append(word)
    else:
        ans[letter]=[word]
print(ans)
```

Given a nested dictionary:
students = {
    "Rahul": {"math": 80, "science": 90},
    "Aman": {"math": 70, "science": 85},
    "Neha": {"math": 95, "science": 92}
}

calculate each student's average and create a new dictionary containing their averages.

```python
students = {
    "Rahul": {"math": 80, "science": 90},
    "Aman": {"math": 70, "science": 85},
    "Neha": {"math": 95, "science": 92}
}
ans={}
for key,value in students.items():
    values=value.values()
    sum_marks=sum(values)
    lenght=len(values)
    average=sum_marks/lenght
    if key not in ans:
        ans[key]={"average":average}
print(ans)
            
```
