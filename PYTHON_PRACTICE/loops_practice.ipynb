Multiplication Table Analyzer

Take a number n and a limit as input.

Print the multiplication table of n from 1 to limit.

Also calculate:

1. Number of even results
2. Number of odd results
3. Sum of all even results
4. Sum of all odd results

```python
even_results=[]
odd_results=[]
sum_even=0
sum_odd=0
n=int(input("Enter the table you want"))
limit=int(input("Enter the limit"))

for i in range(1,limit+1):
    table=i*n
    print(table)
    if i%2==0:
        even_results.append(i)
        sum_even+=i
    if i%2!=0:
        odd_results.append(i)
        sum_odd+=i

print(even_results)
print(sum_even)
print(odd_results)
print(sum_odd)
```

Number Guessing Game 🎯

Create a secret number.

Keep asking the user to guess the number until they guess correctly.

After every guess:

If the guess is too small → Too Low
If the guess is too large → Too High
If correct → Correct!

Finally display the total number of attempts.

```python
i = 55
while True:
    guessing=int(input("Enter your number"))
    if guessing==i:
        print("You guessed the number")
        break
    elif guessing>i:
        print("high")
    elif guessing<1:
        print("low")
```
