Given:

nums = [10, 20, 10, 30, 20, 40, 10]

Create a new list containing each value only once.

```python
nums = [10, 20, 10, 30, 20, 40, 10]
ans=[]
for x in nums:
    if x not in ans:
        ans.append(x)
print(ans)
```

Find the second-largest number in a list without using sort().

```python
a=[1, 2, 3, 4, 5]
largest=0
second_largest=0
for x in a:
    if x > largest:
        second_largest=largest
        largest=x
    elif x >second_largest and x!= largest:
        second_largest=x
print(second_largest)
        
```

Given a list containing integers from 1 to n with one number missing, find the missing number

```python
n=int(input("Enter the n number "))
nums=list(map(int,input(f"enter a list missing of {n} number but missing one int:").split(",")))
print(nums)
excepted_sum=n*(n+1)//2
print(excepted_sum)
real_sum=sum(nums)
print(real_sum)
print(excepted_sum-real_sum)
```

Given:
nums = [1, 2, 2, 3, 4, 4, 5, 5, 5]

create a list containing only values that occur more than once, without using Counter.

```python
nums = [1, 2, 2, 3, 4, 4, 5, 5, 5]
container=[]
ans=[]
for x in nums:
    if x not in container:
        container.append(x)
    elif x in container and x not in ans:
        ans.append(x)
print(ans)
```

Rotate a list to the right by k positions using slicing.

```python
a=[1,2,3,4,5]
d=2
n=len(a)
k=d%n
print(k)
ans=a[-k:]+a[:-k]
print(ans)
```

Given:
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

create a flattened list containing only even numbers using a nested list comprehension.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
ans=[y for x in matrix for y in x if y%2==0]
print(ans)
```
