Given a tuple of numbers, find the maximum and minimum without using max() or min().

```python
t=(1,2,3,4,5,6)
maxv=0
minv=t[0]
for x in t:
    if x>maxv:
        maxv=x
    elif x<minv:
        minv=x
print(maxv)
print(minv)
```

```python
t = (1, 2, 2, 3, 3, 3, 4)
l=[]
for x in t:
    if x not in l:
        l.append(x)
    elif x in l:
        pass
print(l)
ans=tuple(l)
print(ans)
```

Sort a tuple of tuples based on the second element:
(
    ("A", 30),
    ("B", 10),
    ("C", 20)
)

```python
t=(
    ("A", 30),
    ("B", 10),
    ("C", 20)
)
l=[]
for w,v in t:
    l.append([v,w])
l.sort()
lists=[]
for v,w in l:
    lists.append((w,v))
ans=tuple(lists)
print(ans)
```

Given a tuple of student records:
students = (
    ("Rahul", 85),
    ("Aman", 92),
    ("Neha", 78),
    ("Raj", 95)
)

find the student with the highest marks and return their complete tuple.

```python
students = (
    ("Rahul", 85),
    ("Aman", 92),
    ("Neha", 78),
    ("Raj", 95)
)
top=students[0]
for x in students[1:]:
    if x>top:
        top=x
print(top)
```
