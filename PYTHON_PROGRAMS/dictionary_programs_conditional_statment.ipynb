```python
student_id=input("Enter student ID Example student_1").lower()
query=input("Enter your query: 1. Result 2. Scholarship 3. Performace").lower()
minimum_attendenc=int(input("Enter The minimum attendence criteria"))
minimum_average_marks=int(input("Enter The minium average_marks"))
if student_id in students:
    print("Valid Student ID")
    divder=len(students[student_id]["marks"].values())
    average=(sum(students[student_id]["marks"].values()))/divder
    if query=="result":
        print(f"Result of {students[student_id]["name"]} is: {average}")
    elif query=="scholarship":
        if students[student_id]["attendance"]>=85 and students[student_id]["scholarship"]==True and average>=80 and min(students[student_id]['marks'].values())>=50 :
            print(f"Scholarship granted to student{students[student_id]["name"]}")
        else:
            print(f"Scholariship not valid for {students[student_id]["name"]}")
    elif query=="performance":
        if students[student_id]['attendance']>=90 and average>=90:
            print(f"student {students[student_id]["name"]} is an Elite Performer")
        elif students[student_id]['attendance']<75 and average>=80:
            print(f"student {students[student_id]["name"]} is Acadmically Strong but Attendance Risk")
        elif students[student_id]['marks']<40:
            print(f"student {students[student_id]["name"]} is on Academic Risk")
        else:
            print("regular performer")






```
