## Arthemetic operators

```python
a=float(input("Enter a number: "))
b=float(input("Enter another number: "))
print(f"The sum of {a} and {b} is {a+b}")
print(f"The difference of {a} and {b} is {a-b}")
print(f"The product of {a} and {b} is {a*b}")
print(f"The quotient of {a} and {b} is {a/b}")
print(f"The floor division of {a} and {b} is {a//b}")
print(f"The remainder of {a} and {b} is {a%b}")
print(f"The power of {a} and {b} is {a**b}")
```

## Comparision

```python
a=float(input("Enter a number: "))
b=float(input("Enter another number: "))
print(f"are a and b equal? {a==b}")
print(f"are a and b not equal? {a!=b}")
print(f"is a greater than b? {a>b}")
print(f"is a less than b? {a<b}")
print(f"is a greater than or equal to b? {a>=b}")
print(f"is a less than or equal to b? {a<=b}")
```

## augumented

```python
a=float(input("Enter a number: "))
b=float(input("Enter another number: "))
a += b
print(f"the augmented assignment of a and b is {a}")
a -= b
print(f"the augmented assignment of a and b is {a}")
a *= b
print(f"the augmented assignment of a and b is {a}")
a /= b
print(f"the augmented assignment of a and b is {a}")
a //= b
print(f"the augmented assignment of a and b is {a}")
a %= b
print(f"the augmented assignment of a and b is {a}")
a **= b
print(f"the augmented assignment of a and b is {a}")
```

## bitwise

```python
a=int(input("Enter a number: "))
b=int(input("Enter another number: "))
print(f"the bitwise AND of a and b is {a&b}")
print(f"the bitwise OR of a and b is {a|b}")
print(f"the bitwise XOR of a and b is {a^b}")
print(f"the bitwise NOT of a is {~a}")
print(f"the left shift of a by b positions is {a<<b}")
print(f"the right shift of a by b positions is {a>>b}")
```
