# For continue and pass

```python
n=int(input("Enter one "))

even_number=[]
odd_number=[]
even_counter=0
odd_counter=0
for i in range(1,n+1):
    if i%2==0:
        if i==4:
            continue
        even_number.append(i)
        even_counter+=1

    if i%2!=0:
        odd_number.append(i)
        odd_counter+=1
    else:
        pass

print(even_number)
print(even_counter)
print(odd_number)
print(odd_counter)    
```

# While and break

```python
correct_password="mbykhskvkkk"

attempt=3

while attempt>0:
    input_password=str(input("Enter your password"))
    if input_password==correct_password:
        print("Login sucssesfuly")
        break
    else:
        attempt-=1
        if attempt>0:
            print(f"Try again attempt left: {attempt}")
        else:
            print("Account blocked")
```
