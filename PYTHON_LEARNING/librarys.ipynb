# math

```python
import math
print(math.pi)
print(math.e)
print(math.tau)
print(math.sqrt(25))
print(math.pow(5,5))
print(math.factorial(7))
print(math.fabs(-55))
print(math.floor(5.5))
print(math.ceil(5.5))
print(math.sin(math.pi/2))
print(math.radians(180))
print(math.degrees(math.pi))
print(math.log(math.e))
```

# random 

```python
import random
print(random.random())
print(random.randint(1,10))
print(random.randrange(1,11,2))
print(random.choice(["Harsh","19","student"]))
print(random.choices(["Harsh","19","student"],k=2))
```

# re

```python
import re
text="my name is harsh i am 19 years old"
print(re.search(r"\d+",text))
print(re.sub(r"19","20",text))
print(re.findall(r"\D+",text))
text = "123abc"
print(re.match(r"\d+",text))
```
