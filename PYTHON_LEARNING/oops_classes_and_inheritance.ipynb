### classes and objects

```python
class student:
    def __init__(self,name,marks,attandance,scholarship):
        self.name=name
        self.marks=marks
        self.attandance=attandance
        self.scholarship=scholarship
    def report(self):
        print(f"student name is {self.name}")
        print(f"student marks are {self.marks}")
        print(f"student attandance status is {self.attandance}")
        print(f"student scholarship status is {self.scholarship}")
```

```python
student1=student("Harsh","96.7","90.0","Active")
```

```python
student1.report()
```

## Inheritance
### Single

```python
class students:
    def __init__(self,name,student_id,subject_list):
        self.name=name
        self.student_id=student_id
        self.subject_list=subject_list
        self.student_data={}
    def status(self):
        marks={}
        for subject in self.subject_list:
            mark=float(input(f"Enter the {subject} number:"))
            marks[subject]=mark
        self.student_data[self.student_id]={
            "name":self.name,
            "marks":marks
        }
        print(self.student_data)

class persent(students):
    def __init__(self,name,student_id,subject_list):
        super().__init__(name,student_id,subject_list)
    def persentage(self):
        student_persent=sum(self.student_data[self.student_id]["marks"].values())/len(self.student_data[self.student_id]["marks"].values())
        return student_persent
    
```

```python
persent1=persent("Harsh",1,["Math","CS"])
```

```python
persent1.status()
```

```python
persent1.persentage()
```

### multilevel

```python
class students:
    def __init__(self,name,student_id,subject_list):
        self.name=name
        self.student_id=student_id
        self.subject_list=subject_list
        self.student_data={}
    def status(self):
        marks={}
        for subject in self.subject_list:
            mark=float(input(f"Enter the {subject} number:"))
            marks[subject]=mark
        self.student_data[self.student_id]={
            "name":self.name,
            "marks":marks
        }
        print(self.student_data)

class persent(students):
    def __init__(self,name,student_id,subject_list):
        super().__init__(name,student_id,subject_list)
    def persentage(self):
        student_persent=sum(self.student_data[self.student_id]["marks"].values())/len(self.student_data[self.student_id]["marks"].values())
        return student_persent
    
class attandance(persent):
    def __init__(self,name,student_id,subject_list,attandance):
        super().__init__(name,student_id,subject_list)
        self.attandance=attandance
    def attandance_status(self):
        if self.attandance>=75:
            print(f"student {self.student_data[self.student_id]["name"]} appearance is good")
        else:
            print("accadmic risk")
```

```python
student1=attandance("Harsh",1,["Math","CS"],79)
```

```python
student1.status()
```

```python
student1.persentage()
```

```python
student1.attandance_status()
```

### multiple

```python
class students:
    def __init__(self,name,student_id,subject_list):
        self.name=name
        self.student_id=student_id
        self.subject_list=subject_list
        self.student_data={}
    def status(self):
        marks={}
        for subject in self.subject_list:
            mark=float(input(f"Enter the {subject} number:"))
            marks[subject]=mark
        self.student_data[self.student_id]={
            "name":self.name,
            "marks":marks
        }
        print(self.student_data)

class collage:
    def __init__(self,collage_name):
        self.collage_name=collage_name
    def collage_status(self):
        print(f"collage name is {self.collage_name}")

class persent(students,collage):
    def __init__(self,name,student_id,subject_list,collage_name):
        students.__init__(self,name,student_id,subject_list)
        collage.__init__(self,collage_name)
    def persentage(self):
        student_persent=sum(self.student_data[self.student_id]["marks"].values())/len(self.student_data[self.student_id]["marks"].values())
        return student_persent

```

```python
persent1=persent("Harsh",1,["Math","CS"],"ABCD")
```

```python
persent1.status()
```

```python
persent1.collage_status()
```
