## Data Types

```python
a=15
print(f"The value of a is {a}")
print(f"The type of a is {type(a)}")
b=3.14
print(f"The value of b is {b}")
print(f"The type of b is {type(b)}")
c="Hello, World!"
print(f"The value of c is {c}")
print(f"The type of c is {type(c)}")
d=True
print(f"The value of d is {d}")
print(f"The type of d is {type(d)}")
```

## IEEE 754 FORMAT

```python
a=0.1+0.2
b=0.3
c=a==b
print(f"The value of c is {c}")
```

## Str Imutable

```python
a="Hello"
print(f"The id of a is {id(a)}")
b=a+" World!"
print(f"The id of b is {id(b)}")
```

## Bool subclass of int

```python
print(f"Is bool same as int? {isinstance(True, int)}")
print(f"Is bool same as int? {issubclass(bool, int)}")
```
