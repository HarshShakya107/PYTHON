## Encapsulation

### public

```python
class calculate:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def add(self):
        print(self.a+self.b)
```

```python
calculation=calculate(66,87)
calculation.add()
```

```python
dir(calculation)
```

```python
calculation.a
```

### private

```python
class calculate:
    def __init__(self,a,b):
        self.__a=a
        self.__b=b
```

```python
calculation=calculate(5,6)
```

```python
dir(calculation)
```

```python
calculation._calculate__a
```

### getter and setter

```python
class calculate:
    def __init__(self,a,b):
        self.__a=a
        self.__b=b
    @property
    def info(self):
        print(self.__a)
        print(self.__b)
    @info.setter
    def info(self,value_tup):
        new_a,new_b=value_tup
        if self.__a<new_a:
            self.__a=new_a
        if self.__b<new_b:
            self.__b=new_b
```

```python
calculation=calculate(5,6)
```

```python
calculation.info
```

```python
calculation.info=(55,66)
```

```python
calculation.info
```
