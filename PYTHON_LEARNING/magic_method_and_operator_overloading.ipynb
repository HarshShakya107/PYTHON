## magic methods and operator overloading

```python
class book:
    def __init__(self,author,title,price):
        self.author=author
        self.title=title
        self.price=price
    def __str__(self):
        return f"the price of book {self.title}: {self.price}"

    def __add__(self,other):
        return self.price+other.price
```

```python
book1=book("Harsh","Nothing",0)
book2=book("Ujjwal","Nothing",0)
```

```python
print(book1)
```

```python
result=book1+book2
```
