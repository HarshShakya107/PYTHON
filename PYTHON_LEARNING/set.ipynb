## methods

```python
a={1, 2, 3}
a.add(5)
print(a)
a.update({6, 7, 8})
print(a)
a.pop()
print(a)
a.remove(3)
print(a)
a.discard(2)
print(a)
b=a.copy()
print(b)
a.clear()
print(a)
```
