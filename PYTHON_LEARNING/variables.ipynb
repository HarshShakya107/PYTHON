## Dynamic Typing

```python
x=10
print(f"The value of x is {x}")
print(f"the unique identifier of this variable is {id(x)}")
x=20
print(f"The value of x is {x}")
print(f"the unique identifier of this variable is {id(x)}")
```

## Print

```python
print("Hello, Harsh")
# end=" " is used to avoid the new line after the print statement.
print("Hello, Harsh",end=" ")
# sep="-" is used to separate the values with a hyphen instead of a space.
print("1","2","3",sep="-")
```

## Input

```python
input_string = input("Enter a string: ")
print(f"You entered: {input_string}")
print(f"The type of input_string is {type(input_string)}")
```

## Type Casting

1. int()
2. float()
3. str()
4. bool()
5. list()
6. tuple()
7. set()
8. dect()

```python
a=int(input("Enter a number: "))
print(f"The value of a is {a}")
print(f"The type of a is {type(a)}")
b=float(input("Enter a decimal number: "))
print(f"The value of b is {b}")
print(f"The type of b is {type(b)}")
c=bool(input("Enter a boolean value (True/False): "))
print(f"The value of c is {c}")
print(f"The type of c is {type(c)}")
d=list(input("Enter a list of values separated by commas: ").split(","))
print(f"The value of d is {d}")
print(f"The type of d is {type(d)}")
e=tuple(input("Enter a tuple of values separated by commas: ").split(","))
print(f"The value of e is {e}")
print(f"The type of e is {type(e)}")
f=set(input("Enter a set of values separated by commas: ").split(","))
print(f"The value of f is {f}")
print(f"The type of f is {type(f)}")
list_name=[('name','Harsh'),( 'age',25)]
g=dict(list_name)
print(f"The value of g is {g}")
print(f"The type of g is {type(g)}")
```
