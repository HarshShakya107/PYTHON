## exception handling 

### value error

```python
try:
    a=int(input("Enter your numbr"))
    c=1/a
except ValueError as ex:
    print(ex)
else:
    print(c)
finally:
    print("programed runs")
```

### ZeroDivisionError

```python
a=int(input("Enter your numbr"))
try:
    c=1/a
except ZeroDivisionError as ex:
    print(ex)
else:
    print(c)
finally:
    print("programed runs")
```

### name error

### type error

```python
a=int(input("Enter your numbr"))
try:
    b="a"
    c=b/a
except TypeError as ex:
    print(ex)
else:
    print(c)
finally:
    print("programed runs")
```

### File not found

```python
try:
    f=open("practice.txt","r")
except FileNotFoundError as ex:
    print(ex)
else:
    output=f.read()
    print(output)
    f.close()
finally:
    print("programed runs")
```

### exception

```python
try:
    f=open("practice.txt","r")
except Exception as ex:
    print(ex)
else:
    output=f.read()
    print(output)
    f.close()
finally:
    print("programed runs")
```
