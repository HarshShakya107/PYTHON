## dictionary

```python
a={"Name": ["John", "Doe","Harsh"], 
   "Age": [30, 25, 35]}
print(type(a))
print(a)
a={"Name": ["John", "Doe","Harsh"], 
   "Age": [30, 25, 35],"Name": ["Harsh"],}
print(a)
```

## methods

```python
a={"Name": ["John", "Doe","Harsh"], 
   "Age": [30, 25, 35]}
print(a.get("Name"))
print(a.get("Gender","Not Found"))
print(a.keys())
print(a.values())
print(a.items())
a.update({"Gender":"Male"})
print(a)
a.update({"Name":["John","Moe","Harsh","Rohit"]})
print(a)
a.pop("Gender")
print(a)
a.popitem()
print(a)
a.clear()
print(a)
b={"Name": ["John", "Doe","Harsh"], 
   "Age": [30, 25, 35]}
c=b
print(b)
print(c)
b.update({"Gender":"Male"})
print(b)
print(c)
d=b.copy()
print(d)
b.update({"Gender":"Female"})
print(b)
print(d)
```

## nested dictionary

```python
a={"student1":{"Name":"John","Age":30},
   "student2":{"Name":"Doe","Age":25}}
print(a)
print(a.get("student1"))
```

## dictionary comprihension

```python
a={x: x**2 for x in range(20) if x > 2}
print(a)
```

## dictionary merge

```python
a={"Name": ["John", "Doe","Harsh"], "Age": [30, 25, 35]}
b={"Gender": ["Male", "Female", "Male"], "Age": [30, 55, 35]}
c={**a,**b}
print(c)
```
