# write multiple lines

```python
lists=["\n my name is Harsh","\n its practice"]
with open("C:\\Users\\HP\\OneDrive\\Desktop\\GEN_AI_LEARNING\\PYTHON\\Learning\\JUNK_FOLDER\\practice_file.txt","w+") as f:
    f.writelines(lists)
    f.seek(0)
    output=f.read()
    print(output)
```

```python
with open("C:\\Users\\HP\\OneDrive\\Desktop\\GEN_AI_LEARNING\\PYTHON\\Learning\\JUNK_FOLDER\\practice_file.txt","r") as f:
    lines=f.readlines()
    total_lines=len(lines)
    total_words=sum(len(line.split()) for line in lines)
    total_chars=sum(len(line) for line in lines)
    print(total_lines,total_chars,total_words)
```
